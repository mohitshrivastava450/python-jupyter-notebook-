```python
n=int(input("enter any number"))
for i in range(n):
    for j in range(i):
        print(" ",end=" ")
    for j in range(n-1-i):
        print("*",end=" ")
    for j in range(n-i):
        print("*",end=" ")
    print()
```

```python
n=int(input("enter any number"))
for i in range(n-1):
    for j in range(n-i-1):
        print(" ",end=" ")
    for j in range(i+1):
        print("*",end=" ")
    for j in range(i):
        print("*",end=" ")
    print()
for k in range(n):
    for l in range(k):
        print(" ",end=" ")
    for l in range(n-1-k):
        print("*",end=" ")
    for l in range(n-k):
        print("*",end=" ")
    print()
```

```python
n=int(input("enter any number"))
for i in range(n):
    for j in range(n-i-1):
        print(" ",end=" ")
    for j in range((2*i)+1):
        if j==0 or j==2*i or i==n-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
```
