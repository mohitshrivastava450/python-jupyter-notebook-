```python
l=[]
while True:
    ch=int(input('''
    1.push
    2.pop
    3.peek
    4.display stack
    5.exit
    '''))
    if ch==1:
        n=input("enter any number")
        l.append(n)
        print(l)
    elif ch==2:
        if len(l)==0:
            print("empty stack")
        else:
            p=l.pop()
            print(p)
            print(l)
    elif ch==3:
        if len(l)==0:
            print("empty stack")
        else:
            print("last element of stack: ",l[-1])
    elif ch==4:
        print("display stack: ",l)
    elif ch==5:
        break
    else:
        print("invalid choice ")
```

```python
l=[]
while True:
    ch=int(input('''
    1.Enqueue
    2.Dequeue
    3.Front Element
    4.Rear Element
    5.Display
    6.Exit
    '''))
    if ch==1:
        n=input("enter any number")
        l.append(n)
        print(l)
    elif ch==2:
        if len(l)==0:
            print("empty queue ")
        else:
            p=l.pop()
            print(p)
            print(l)
    elif ch==3:
        if len(l)==0:
            print("empty queue ")
        else:
            print("1st element of queue: ",l[0])
    elif ch==4:
        if len(l)==0:
            print("empty queue ")
        else:
            print("Rear element of queue: ",l[-1])
    elif ch==5:
        print("display queue : ",l)
    elif ch==6:
        break
    else:
        print("invalid choice")
```

### list comprehension

```python
l=[1,2,3,4,5]
l1=[]
for i in l:
    l1.append(i**3)
print(l1)
#[expression loop condition]
print([i**3 for i in l]) 
```

```python
l1=[5,1,3,4,2]
l2=[2,5,1,3,4]
l3=[]
for i,j in zip(l1,l2):
    l3.append(i**j)
print(l3)
print([i**j for i,j in zip(l1,l2)])
```
