# set 

```python
s={}
s=set()
print(type(s))
```

```python
s={10,12.5,"hello",100} #heterogenous
print(s)      #unordered, unindexed
#add #pop
s.add(120)              #mutable
print(s)
for i in s:             #direct iteration
    print(i)
print(s.pop())
print(s)
print(s.pop())
print(s)
```

# remove

```python
s={10,12.5,"hello",100}
print(s)
print(s.remove(10))
print(s)
print(s.remove("hello"))
print(s)
```

# clear

```python
s={10,12.5,"hello",100}
print(s)
s.clear()
print(s)
```

# discard

```python
s={10,12.5,"hello",100} 
print(s)   
s.discard(10)
print(s)
```

# copy

```python
s={10,12.5,"hello",100}
print(s)   
a=s.copy()
print(a)
```

# union

```python
s={10,12.5,"hello",100}
t={10,15,"hello",120,132}
print(s)   
print(s.union(t))
```

# intersection

```python
s={10,12.5,"hello",100}
t={10,15,"hello",120,132}
print(s)   
print(s.intersection(t))
```

# difference

```python
s={10,12.5,"hello",100}
t={10,15,"hello",120,132}
print(s)   
print(s.difference(t))
```

# symmetric_difference

```python
s={10,12.5,"hello",100}
t={10,15,"hello",120,132}
print(s)   
print(s.symmetric_difference(t))
```
