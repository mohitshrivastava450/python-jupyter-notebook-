```python
#wap function that returns whether a given number is prime or not
def  prime(n):
    if n==0 or n==1:
        return False
    elif n>1:
        for i in range(2,n):
            if n%i==0:
                return False
                break
        else:
                return True
    else:
        return False
```

```python
num=int(input("Enter any number : "))
if prime(num):
    print(num,"prime number")
else:
    print(num,"not a prime")
```

```python
#wapf that returns armstrong no. between 1 to 2000
def armstrong():
    l=[]
    for i in range(1,2001,1):
        if i>=1 and i<10:
            l.append(i)
        elif i<100:
            unit=i%10
            ten=i-unit
            if i==unit**2+ten**2:
                l.append(i)
        elif i<1000:
            unit=i%10
            ten=i-i%10
            hundred=(i-i%100)
            ten=(ten-hundred)//10
            hundred=(i-i%100)//100
            if i==unit**3+ten**3+hundred**3:
                l.append(i)
        else:
            unit=i%10
            ten=i-i%10
            hundred=i-i%100
            ten=(ten-hundred)//10
            thousand=(i-i%1000)
            hundred=(hundred-thousand)//100
            thousand=(i-i%1000)//1000
            if i==unit**4+ten**4+hundred**4+thousand**4:
                l.append(i)
    return l
```

```python
def arms(start,end):
     l=[]
     for i in range(start,end+1):
         if i>0:
             a=len(str(i))
             sm=0
             arm=i
             while i!=0:
                 dig=i%10
                 sm=sm+(dig**a)
                 i=i//10
             if arm==sm:
                 l.append(arm)
     return l
```

```python
#wapf that return whether a given number is armstrong or not
def arm_strong(n):
    a=len(str(n))
    sm=0
    arm=n
    while n!=0:
        dig=n%10
        sm=sm+(dig**a)
        n=n//10
    return sm==arm
```

```python
num=int(input("Enter any number: "))
if arm_strong(num):
    print(num,"armstrong")
else:
    print(num,"not an armstrong")
```

```python
#wapf that return a fibonacci series upto n term
def fibo():
    n=int(input("enter a number: "))
    if n==1:
        return [0]
    elif n==2:
        return [0,1]
    fib=[0,1]
    for i in range(2,n):
            next_term=fib[-1]+fib[-2]
            fib.append(next_term)
    return fib
```

```python
#wapf that return nth term of fibonacci
def fibonacci():
    n=int(input("enter a number: "))
    if n==1:
        return [0]
    elif n==2:
        return [0,1]
    fib=[0,1]
    for i in range(2,n):
            next_term=fib[-1]+fib[-2]
            fib.append(next_term)
    return fib[-1]
    
```

```python
#wapf that return whether a given string is palendrom or not
def palindrome():
    n=input("enter a word : ")
    rev=""
    for i in n:
        rev=i+rev
    if n==rev:
        print(n,"palindrome")
    else:
        print(n,"not a palindrome")
    
    
```

```python
#wapf that takes a list and return a new list with distinct element
#from 1st list
#l=[1,1,2,10,2,5,10,3,4]
def distlist(l):
    l1=[]
    for i in l:
        if i not in l1:
            l1.append(i)
    return l1
```

```python
def sort_list(l,reverse=False):
    if reverse==False:
        for i in range(i+1,len(l)):
            for j in range(i+1,len(l)):
                if l[i]>l[j]:
                    l[i],l[j]=l[j],l[i]
        return l
    else:
        for i in range(len(l)):
            for j in range(i+1,len(l)):
                if l[i]<l[j]:
                    l[i],l[j]=l[j],l[i]
        return l
```

```python
l1=[10,2,100,19]
sort_list(l1,reverse=True)
```

```python
# wapf that returns mean,median and mode from a given parameter(take list as a parameter)
def mean_median_mode(l):
    sum=0
    count=0
    for i in l:
        sum=sum+i
        count=count+1
    mean=sum/count
    sorted_list=sorted(l)
    n=len(l)
    if n%2==0:
        median=(sorted_list[n//2]+sorted_list[n//2-1])/2
    else:
        median=sorted_list[n//2]
    d={}
    for i in l:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    max_value=max(d.values())
    for i,j in d.items():
        if j==max_value:
            mode=i
    return mean,median,mode
    
    
```

```python
mean_median_mode([1,2,3,4,5,6,7,5])
```

```python
# high order function
# map(function,iterables)
l=[1,2,3,4,5]
# output- [1,8,27,64,125]
l1=[]
for i in l:
    l1.append(i**3)
print(l1)                          #simple
print([i**3 for i in l])           # with list comprehension
print(list(map(lambda i:i**3,l)))  # with high order function
```
